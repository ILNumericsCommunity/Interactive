# Welcome to the `ILNumerics.Community.Interactive` Notebook

### *This package adds math support and visualizations for ILNumerics projects.*

## 1. Build the project


```polyglot-notebook
# Clear any older versions of this extension package from your NuGet cache
#rm ~/.nuget/packages/ILNumerics.Community.Interactive -Force -Recurse -ErrorAction Ignore
#rm $env:NUGET_PACKAGES\ILNumerics.Community.Interactive -Force -Recurse -ErrorAction Ignore

# Build the project
dotnet build Interactive

# Pack up the NuGet package. 
dotnet pack Interactive /p:PackageVersion=7.0.0 -c Debug

# 3. Check that the package is there
Get-ChildItem -Recurse ILNumerics.Community.Interactive*.nupkg
```

## 2. Load the NuGet package 

Now we're ready to install the extension that's packed in the NuGet package we just built.

First, let's make sure the file is there like we expect after the build.


```polyglot-notebook
using System.IO;

// Create an absolute path since #i doesn't like 
var debugOutputDirInfo = new DirectoryInfo(@".\");
var debugOutputPackagePath = debugOutputDirInfo.EnumerateFiles("ILNumerics.Community.Interactive.7.*.nupkg").FirstOrDefault()?.FullName;

if (debugOutputPackagePath != null && File.Exists(debugOutputPackagePath))
    $"✅ Package {Path.GetFileName(debugOutputPackagePath)} is available!".Display();
else
    "❌ Something must have gone wrong with the build. The package isn't available.".Display();
```

If the package is there, we can include its location as a NuGet source using the `AddPackageSource` command.

```polyglot-notebook
using Microsoft.DotNet.Interactive;
using Microsoft.DotNet.Interactive.Commands;

await Kernel.Root.FindKernelByName("csharp").SendAsync(new AddPackageSource(debugOutputDirInfo.FullName));
```
